```python
import json

for (dataset, model), indices in selected_indices.items():
    with open(f'data/collection/{model}/{dataset}.json') as f:
        data = json.load(f)
    selected_data = [data[i] for i in indices]
    with open(f'data/selection/{model}/{dataset}.json', 'w+') as f:
        json.dump(selected_data, f, indent=2)
```
